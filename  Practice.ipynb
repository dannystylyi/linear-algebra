```python
# Упражнение 2.4

import matplotlib.pyplot as plt
import numpy as np

vec_1 = np.array(list(map(float, input().split())))


def modulation(vector, norm):
    a = np.linalg.norm(vector)
    unit_norm = norm * (vector / a)
    plt.arrow(*unit_norm, color='g', width=0.07)
    plt.arrow(*vector, color='b', alpha=0.3, width=0.07)
    plt.show()
    return unit_norm


a = modulation(vec_1, 9)
print(a, np.linalg.norm(a), sep='\n')
```

```python
# Упражнение 4.2
import matplotlib.pyplot as plt
import numpy as np


def pirs__and__cos_equal(vek1, vek2):
    vek1_cert = sum(vek1) / len(vek1)
    vek2_cert = sum(vek2) / len(vek2)
    v1 = np.copy(vek1)
    v2 = np.copy(vek2)
    cos_vek12 = np.dot(v1, v2) / (np.dot(v1, v1) ** .5 * np.dot(v2, v2) ** .5)
    for i in range(len(vek1)):
        v1[i] -= vek1_cert
    for i in range(len(vek2)):
        v2[i] -= vek2_cert
    ro = np.dot(v1, v2) / (np.dot(v1, v1) ** .5 * np.dot(v2, v2) ** .5)
    return ro, cos_vek12, vek2_cert


variable = (-50, 50)
a = np.array(np.arange(4))
plot_ro = np.zeros(100)
plot_cos_vek12 = np.zeros(100)
centered = np.zeros(100)

for i, j in enumerate(range(variable[0], variable[1])):
    b = np.array(np.arange(4)) + j
    plot_ro[i], plot_cos_vek12[i], centered[i] = pirs__and__cos_equal(a, b)

plt.plot(np.arange(-50, 50), plot_ro, 'b-', linewidth=2, label='Pirson')
plt.plot(centered, plot_cos_vek12, 'r--', linewidth=2, label='cos(v1^v2)')
plt.title('Корреляция Пирсона и косинусное сходство ')
plt.xlabel('Сдвиг вектора b' )
plt.ylabel('Значение метрики' )

plt.legend()
plt.show()
```

```python
# Упражнение 4.4
import numpy as np
import time


def pirs__and__cos_equal(vek1, vek2):
    vek1_cert = sum(vek1) / len(vek1)
    vek2_cert = sum(vek2) / len(vek2)
    for i in range(len(vek1)):
        vek1[i] -= vek1_cert
    for i in range(len(vek2)):
        vek2[i] -= vek2_cert
    ro = np.dot(vek1, vek2) / (np.dot(vek1, vek1) ** .5 * np.dot(vek2, vek2) ** .5)
    return ro


def rho(x, y):
    xm = x - np.mean(x)
    ym = y - np.mean(y)
    n = np.dot(xm, ym)
    d = np.linalg.norm(xm) * np.linalg.norm(ym)
    return n / d


t1 = time.time()
for x in range(1000):
    i = np.random.randn(50000, 2)
    pirs__and__cos_equal(i[0], i[1])
t2 = time.time()
t_1 = t2 - t1

t3 = time.time()
for x in range(1000):
    i = np.random.randn(50000, 2)
    np.corrcoef(i[0], i[1])
t4 = time.time()
t_2 = t4 - t3
print(round(t_1 / t_2 * 100, 3), round(t_2 - t_1, 3))
```

```python
# Упражнение 4.5

import numpy as np
import matplotlib.pyplot as plt

core = np.array([-1, 2])
signal_y = [0] * 13 + [10] * 14 + [0] * 12

plt.figure(figsize=(10, 5))
plt.subplot(1, 3, 1)
plt.plot(core, 'b-', linewidth=2)
plt.title('Ядро')

plt.subplot(1, 3, 2)
plt.plot(signal_y, 'r-', linewidth=2)
plt.title('Сигнал')

dif = []
for x in range(len(signal_y) - 1):
    if signal_y[x] - signal_y[x + 1] != 0:
        print(np.dot(core, signal_y[x: x+2]))
        dif.append(abs(np.dot(core, signal_y[x: x+2])) / np.dot(core, signal_y[x: x+2]))
    else:
        dif.append(0)
plt.subplot(1, 3, 3)
plt.plot(signal_y, 'r-', linewidth=2, label='Signal')
plt.plot(dif, 'k-.', linewidth=2, label='subtract')
plt.yticks(np.arange(-2, 10.1, 1))

plt.legend()
plt.show()
```

```python
# Упражнение 4.6
# define the kernel (a sorta-kinda Gaussian)
kernel = np.array([0,.1,.3,.8,1,.8,.3,.1,0])
kernel = kernel / np.sum(kernel)

# some handy length parameters
Nkernel = len(kernel)
halfKrn = Nkernel//2


# and the signal
Nsignal = 100
timeseries = np.random.randn(Nsignal)


# plot them
_,axs = plt.subplots(1,2,figsize=(12,4))
axs[0].plot(kernel,'ks-')
axs[0].set_title('Kernel')
axs[0].set_xlim([-1,Nsignal])

axs[1].plot(timeseries,'ks-')
axs[1].set_title('Time series signal')

plt.show()
filtsig = timeseries.copy()

# loop over the signal time points
for t in range(halfKrn+1,Nsignal-halfKrn):
  filtsig[t] = np.dot(kernel,timeseries[t-halfKrn-1:t+halfKrn])


# and plot
_,axs = plt.subplots(1,2,figsize=(12,4))
axs[0].plot(kernel,'ks-')
axs[0].set_title('Kernel')
axs[0].set_xlim([-1,Nsignal])

axs[1].plot(timeseries,color='k',label='Original',linewidth=1)
axs[1].plot(filtsig,'--',color=[.6,.6,.6],label='Smoothed',linewidth=2)
axs[1].legend()

plt.show()
```

```python
# Упражнение 5.1
import numpy as np
import matplotlib.pyplot as plt


# Create the matrix
C = np.arange(100).reshape((10,10))

# extract submatrix
C_1 = C[0:5:1,0:5:1]

# here's what the matrices look like
print(C), print(' ')
print(C_1)



# visualize the matrices as maps
_,axs = plt.subplots(1,2,figsize=(10,5))

axs[0].imshow(C,cmap='gray',origin='upper',vmin=0,vmax=np.max(C))
axs[0].plot([4.5,4.5],[-.5,9.5],'w--')
axs[0].plot([-.5,9.5],[4.5,4.5],'w--')
axs[0].set_title('Original matrix')
# text labels
for (j,i),num in np.ndenumerate(C):
  axs[0].text(i,j,num,color=[.8,.8,.8],ha='center',va='center')


axs[1].imshow(C_1,cmap='gray',origin='upper',vmin=0,vmax=np.max(C))
axs[1].set_title('Submatrix')
# text labels
for (j,i),num in np.ndenumerate(C_1):
  axs[1].text(i,j,num,color=[.8,.8,.8],ha='center',va='center')


plt.show()
```

```python
# Упражнение 5.2

C = np.arange(100).reshape((10,10))
# cut it into blocks
C_1 = C[0:5:1,0:5:1]
C_2 = C[0:5:1,5:10:1]
C_3 = C[5:10:1,0:5:1]
C_4 = C[5:10:1,5:10:1]

# rearrange the blocks
newMatrix = np.vstack( (np.hstack((C_4,C_3)),
                        np.hstack((C_2,C_1))) )

print(newMatrix)
# visualize the matrices
_,axs = plt.subplots(1,2,figsize=(10,5))

axs[0].imshow(C,cmap='gray',origin='upper',vmin=0,vmax=np.max(C))
axs[0].plot([4.5,4.5],[-.5,9.5],'w--')
axs[0].plot([-.5,9.5],[4.5,4.5],'w--')
axs[0].set_title('Original matrix')
# text labels
for (j,i),num in np.ndenumerate(C):
  axs[0].text(i,j,num,color=[.8,.8,.8],ha='center',va='center')


axs[1].imshow(newMatrix,cmap='gray',origin='upper',vmin=0,vmax=np.max(C))
axs[1].plot([4.5,4.5],[-.5,9.5],'w--')
axs[1].plot([-.5,9.5],[4.5,4.5],'w--')
axs[1].set_title('Block-shifted')
# text labels
for (j,i),num in np.ndenumerate(newMatrix):
  axs[1].text(i,j,num,color=[.8,.8,.8],ha='center',va='center')


plt.show()
```
